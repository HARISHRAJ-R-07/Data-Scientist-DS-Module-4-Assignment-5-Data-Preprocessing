####**● Retain Outliers:**
 Retain outliers in the AGE, HbA1c, and BMI columns, explaining why
they should be preserved.
####**● Filter Outliers:**
 In the Creatinine Ratio (Cr) column, exclude values that exceed the
99.5th percentile threshold. Similarly, in the Urea column, exclude values that exceed the 99.9th percentile threshold. Use these specific percentile thresholds for filtering,and do not apply IQR or other methods.
####**● Remove Extreme Outliers:**
 For lipid-related columns (LDL, VLDL, HDL, TG, Chol), remove

```python
# Retain Outliers:
# Retain outliers in the AGE, HbA1c, and BMI columns, explaining why they should be preserved.

sns.boxplot(data=df[["AGE","HbA1c","BMI"]])
plt.show()
```

```python
sns.boxplot(data=df[["Urea","Cr"]])
plt.show()
```

```python
# Filter Outliers:
# In the Creatinine Ratio (Cr) column, exclude values that exceed the 99.5th percentile threshold. Similarly, in the Urea column, exclude values that exceed the 99.9th percentile threshold. Use these specific percentile thresholds for filtering,and do not apply IQR or other methods.

Cr_percentile = df["Cr"].quantile(0.995)
Urea_percentile = df["Urea"].quantile(0.999)
print("Creatinine Ratio(99.5 threshold value): ",Cr_percentile)
print("Urea(99.9 threshold value):",Urea_percentile)
```

```python
df = df[(df["Cr"] <= Cr_percentile) & (df["Urea"] <= Urea_percentile)]
display(df)
```

```python
sns.boxplot(data=df[["Urea","Cr"]])
plt.show()
print(df.describe())
```

```python
# Remove Extreme Outliers:

plt.figure(figsize=(11,7))
sns.boxplot(data=df[["LDL","VLDL","HDL","TG","Chol"]])
plt.show()
```

```python
# For lipid-related columns (LDL, VLDL, HDL, TG, Chol), remove extreme outliers using an appropriate method such as IQR or Z-score.

Q3 = df[["LDL","VLDL","HDL","TG","Chol"]].quantile(0.75)
Q1 = df[["LDL","VLDL","HDL","TG","Chol"]].quantile(0.25)
print("Q3: \n",Q3)
print("Q1: \n",Q1)

IQR = Q3-Q1
print("IQR: \n",IQR)

LDL_VLDL_HDL_TG_Chol_LB = Q1 - 1.5*IQR
LDL_VLDL_HDL_TG_Chol_UB = Q3 + 1.5*IQR
print("Lower bound: \n",LDL_VLDL_HDL_TG_Chol_LB)
print("Upper bound: \n",LDL_VLDL_HDL_TG_Chol_UB)

df = df[((df[["LDL","VLDL","HDL","TG","Chol"]] < (LDL_VLDL_HDL_TG_Chol_LB)).any(axis=1) |
        (df[["LDL","VLDL","HDL","TG","Chol"]] > (LDL_VLDL_HDL_TG_Chol_UB)).any(axis=1))]
```

```python
# Feature Engineering:
# Apply Label Encoding or One-Hot Encoding to the categorical column Gender to convert it into numeric format for model-building purposes.

from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler

LE = LabelEncoder()
df["Gender"] = LE.fit_transform(df["Gender"])
print("Classes in gender column:", list(LE.classes_))
df["Gender"].head(7)
```

```python
ONE = OneHotEncoder(handle_unknown="ignore",sparse_output=False)
df_one_encoded_CLASS = ONE.fit_transform(df[["CLASS"]])
print(df_one_encoded_CLASS)
```

```python
df_one_encoded_CLASS_DF = pd.DataFrame(
    df_one_encoded_CLASS,
    columns=ONE.get_feature_names_out(["CLASS"]),
    index=df.index
)
```

```python
df_one_encoded_CLASS_DF
```

```python
# Feature Scaling:     Standardize or normalize the numerical columns that have varying
# scales (e.g., Patients, AGE, BMI, Cr) to ensure consistent contribution in subsequent
# analyses. Justify the scaling method used (standardization vs. normalization).

std_scaler = StandardScaler()

df_std_scaled = std_scaler.fit_transform(df[['Patient_ID','AGE','BMI','Cr']])
df_std_scaled = pd.DataFrame(
    df_std_scaled,
    columns=['Patient_ID','AGE','BMI','Cr'],
    index=df.index
)
```
